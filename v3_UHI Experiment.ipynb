```python
# Extract average satellite band values in a focal buffer

def extract_focal_buffer(tiff_path, csv_path, buffer_size):
    
    # Load the GeoTIFF data
    data = rxr.open_rasterio(tiff_path)
    tiff_crs = data.rio.crs
    
    # Read the Excel file using pandas
    df = pd.read_csv(csv_path)
    latitudes = df['Latitude'].values
    longitudes = df['Longitude'].values

    # 3. Convert lat/long to the GeoTIFF's CRS
    # Create a Proj object for EPSG:4326 (WGS84 - lat/long) and the GeoTIFF's CRS
    proj_wgs84 = Proj(init='epsg:4326')  # EPSG:4326 is the common lat/long CRS
    proj_tiff = Proj(tiff_crs)
    
    # Create a transformer object
    transformer = Transformer.from_proj(proj_wgs84, proj_tiff)

    # B01_values = []
    B02_values = []
    B03_values = []
    B04_values = []
    B05_values = []
    B06_values = []
    B07_values = []
    B08_values = []
    B8A_values = []
    # B09_values = []
    B11_values = []
    B12_values = []


# Iterate over the latitudes and longitudes, and extract the corresponding mean band values in a the buffer zone
    for lat, lon in tqdm(zip(latitudes, longitudes), total=len(latitudes), desc="Mapping values"):
    # Assuming the correct dimensions are 'y' and 'x' (replace these with actual names from data.coords)
    
        # B01_value = data.sel(
        #     x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
        #     y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
        #     band=1).values
        # B01_value_mean = B01_value.mean()
        # B01_values.append(B01_value_mean)

        B02_value = data.sel(
        x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
        y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
        band=2).values
        B02_value_mean = B02_value.mean()
        B02_values.append(B02_value_mean)

        B03_value = data.sel(
        x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
        y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
        band=3).values
        B03_value_mean = B03_value.mean()
        B03_values.append(B03_value_mean)

        B04_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=4).values
        B04_value_mean = B04_value.mean()
        B04_values.append(B04_value_mean)

        B05_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=5).values
        B05_value_mean = B05_value.mean()
        B05_values.append(B05_value_mean)

        B06_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=6).values
        B06_value_mean = B06_value.mean()
        B06_values.append(B06_value_mean)

        B07_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=7).values
        B07_value_mean = B07_value.mean()
        B07_values.append(B07_value_mean)

        B08_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=8).values
        B08_value_mean = B08_value.mean()
        B08_values.append(B08_value_mean)

        B8A_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=9).values
        B8A_value_mean = B8A_value.mean()
        B8A_values.append(B8A_value_mean)

        # B09_value = data.sel(
        #     x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
        #     y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
        #     band=10).values
        # B09_value_mean = B09_value.mean()
        # B09_values.append(B09_value_mean)
        
        B11_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=11).values
        B11_value_mean = B11_value.mean()
        B11_values.append(B11_value_mean)

        B12_value = data.sel(
            x=slice(lon - buffer_size/111320, lon + buffer_size/111320),
            y=slice(lat + buffer_size/111320, lat - buffer_size/111320),  
            band=12).values
        B12_value_mean = B12_value.mean()
        B12_values.append(B12_value_mean)
    

    # Create a DataFrame to store the band values
    df = pd.DataFrame()
    # df[f'B01_{buffer_size}'] = B01_values
    df[f'B02_{buffer_size}'] = B02_values
    df[f'B03_{buffer_size}'] = B03_values
    df[f'B04_{buffer_size}'] = B04_values
    df[f'B05_{buffer_size}'] = B05_values
    df[f'B06_{buffer_size}'] = B06_values
    df[f'B07_{buffer_size}'] = B07_values
    df[f'B08_{buffer_size}'] = B08_values
    df[f'B8A_{buffer_size}'] = B8A_values
    # df[f'B09_{buffer_size}'] = B09_values
    df[f'B11_{buffer_size}'] = B11_values
    df[f'B12_{buffer_size}'] = B12_values
    
    return df
```

## Remove duplicates

```python
# Make a copy of big_data2 for modelling purposes.

uhi_data = big_data2.copy()
```

```python
print(list(uhi_data.columns[2:-1]))
```

```python
# Remove duplicate rows from the DataFrame based on specified columns and keep the first occurrence

columns_to_check = (list(uhi_data.columns[2:-1]))
for col in columns_to_check:
    # Check if the value is a numpy array and has more than one dimension
    uhi_data[col] = uhi_data[col].apply(lambda x: tuple(x) if isinstance(x, np.ndarray) and x.ndim > 0 else x)

# Now remove duplicates
uhi_data = uhi_data.drop_duplicates(subset=columns_to_check, keep='first')
uhi_data.shape
```

```python
# Resetting the index of the dataset

uhi_data=uhi_data.reset_index(drop=True)
```

## Model Building

```python
# Exclude the latitude and longitude columns

uhi_data = uhi_data.drop(columns=['Longitude', 'Latitude'])
uhi_data.head()
```

### Train and Test Split 

```python
# Split the data into features (X) and target (y), and then into training and testing sets

X = uhi_data.drop(columns=['UHI Index'])
y = uhi_data ['UHI Index']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3,random_state=123)
```

### Feature Scaling 

```python
# Scale the training and test data using standardscaler

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

### Model Training

```python
# Train the Random Forest model on the training data

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train,y_train)
```

## Model Evaluation

### In-Sample Evaluation


```python
# Make predictions on the training data

insample_predictions = model.predict(X_train)
```

```python
# calculate R-squared score for in-sample predictions

Y_train = y_train.tolist()
r2_score(Y_train, insample_predictions)
```

### Out-Sample Evaluation

```python
# Make predictions on the test data

outsample_predictions = model.predict(X_test)
```

```python
# calculate R-squared score for out-sample predictions

Y_test = y_test.tolist()
r2_score(Y_test, outsample_predictions)
```

```python
# Examine importances of features used in the model.

importances = model.feature_importances_

columns_list = list(uhi_data.columns[:-1])
feature_importance = pd.DataFrame()
feature_importance["features"] = columns_list
feature_importance["importance"] = importances
feature_importance = feature_importance.sort_values(by='importance', ascending=False)
print(feature_importance.head(20))
```
